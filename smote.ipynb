```python
import pandas as pd
import numpy as np
from imblearn.over_sampling import SMOTE
from numpy import where
from collections import Counter
import random
```

```python
data = np.loadtxt("data.csv", delimiter=",")
labels = np.loadtxt("data_labels.csv", delimiter=",").astype(int).squeeze()

print("Data shape:", data.shape)
print("Labels shape:", labels.shape)
```

```python
print("Before SMOTE:", np.bincount(labels))
```

## Applying SMOTE
To balance the dataset, we must increase the size of the undersampled class.

```python
smote = SMOTE(random_state=42, sampling_strategy='auto', k_neighbors=5)
X, y = smote.fit_resample(data, labels)
```

```python
print("After SMOTE:", np.bincount(y))
```

## Saving new Dataset
Using a random seed of 42, and getting 800 for the validation set, then save the training set and validation set

```python
rng = np.random.default_rng(42)
N = len(y)

perm = rng.permutation(N)
val_idx = perm[:800]
train_idx = perm[800:]
```

```python
X_train, y_train = X[train_idx], y[train_idx]
X_val,   y_val   = X[val_idx],  y[val_idx]
```

```python
pd.DataFrame(X_train).to_csv("smote/training_set.csv", index=False, header=False)
pd.DataFrame(y_train).to_csv("smote/training_labels.csv", index=False, header=False)
pd.DataFrame(X_val).to_csv("smote/validation_set.csv", index=False, header=False)
pd.DataFrame(y_val).to_csv("smote/validation_labels.csv", index=False, header=False)
```

```python
def counts(arr):
    u, c = np.unique(arr, return_counts=True)
    return dict(zip(u.tolist(), c.tolist()))

print("\n== SUMMARY ==")
print("Training X shape:", X_train.shape, "y shape:", y_train.shape)
print("Validation X shape:", X_val.shape, "y shape:", y_val.shape)
print("Training counts:", counts(y_train))
print("Validation counts:", counts(y_val))
print("Files written:",
      "training_set.csv, training_labels.csv, validation_set.csv, validation_labels.csv", sep="\n- ")
```
